## Etapa 3: Cómo se generaron los datos

**Objetivo:** entender el proceso de captura, porque explica varias características (y problemas) de los datos.

**1. Método: web scraping.** El *scraping* es un programa que abre páginas web automáticamente y copia lo que se ve en pantalla a una tabla. Aquí, el programa del autor hizo búsquedas en amazon.com, recorrió las páginas de resultados y guardó, por cada producto que aparecía, lo que mostraba la ficha resumida del listado. Acá hay un riesgo relevante ya que al utilizar este método, la data se colecta con formato texto (str), lo cual dificulta su procesamiento y limpieza.

**2. Qué representa cada fila.** Una fila **no es un producto**, sino **una aparición de un producto en una página de resultados, en un momento dado**. Si el mismo producto apareció en varias búsquedas o como anuncio patrocinado, quedó registrado varias veces. Por eso las 42.675 filas corresponden a solo 8.808 títulos distintos. Esto es importante para poder dimensionar y entender cómo está compuesto el dataset.

**3. Calendario de captura.** La captura se hizo en **6 días** entre el **21 y el 30 de agosto de 2025** (el detalle por día se calcula en la Etapa 5).

**4. Consecuencias para el análisis** (se tratan en la limpieza):

| Característica de la captura | Efecto en los datos |
|---|---|
| Se copia el texto tal como se ve en pantalla | Todo llega como texto: `"4.6 out of 5 stars"`, `"$1,099.00"`, `"10K+ bought in past month"` |
| Amazon muestra cifras redondeadas | Las compras del último mes vienen en tramos (`50+`, `1K+`), no como número exacto |
| La ficha del listado no siempre trae todos los datos | Vacíos en precio (27,5%), fecha de entrega (27,5%) o botón de compra (34,3%) |
| El texto a veces cae en la columna equivocada | Ej.: `"List Price:"` o `"Typical:"` dentro de la columna de compras |
| Los anuncios patrocinados aparecen en muchas búsquedas | El mismo producto se repite muchas veces, con URL distinta en cada aparición |
| No hay identificador de producto ni categoría | El ASIN se extrae de la URL y la categoría se asigna por palabras clave del título |

**5. Limitaciones y sesgos a tener presentes:**
- **Foto de un momento:** precios, reseñas y distintivos cambian a diario; los datos describen agosto de 2025, no el comportamiento permanente del mercado.
- **Sesgo de visibilidad:** solo entran los productos que Amazon mostró en las búsquedas realizadas; los productos poco visibles están subrepresentados.
- **Búsquedas desconocidas:** no se conocen las palabras clave usadas por el autor, así que la mezcla de categorías refleja sus búsquedas y no el catálogo completo de Amazon.
- **Ruido no electrónico:** aparecen planes de protección, artículos de oficina o membresías, que Amazon mezcla en los resultados.
- **Calidad de las reseñas:** el rating y la cantidad de reseñas los publica Amazon y pueden incluir reseñas no genuinas; no hay forma de verificarlo con estos datos.

## Etapa 5: Verificación de integridad

**Objetivo:** confirmar que el archivo es exactamente el que se usó en el proyecto antes de analizarlo.

### Paso 3 -- Controles del archivo
- **Huella MD5:** es un código calculado a partir de todo el contenido del archivo. Si cambia un solo carácter, cambia la huella. Comparar la huella permite asegurar que el archivo es idéntico al usado en el proyecto.
- **Tamaño, filas y columnas:** deben coincidir con la ficha técnica.
- **Nombres de columnas:** deben ser exactamente los 16 esperados, en el mismo orden, porque los notebooks siguientes los usan por nombre.

Si algún control dice `REVISAR`, el archivo no es el mismo (otra versión del dataset o un archivo editado) y los resultados de los notebooks siguientes pueden variar.

```python
COLUMNAS = ["title", "rating", "number_of_reviews", "bought_in_last_month", "current/discounted_price",
            "price_on_variant", "listed_price", "is_best_seller", "is_sponsored", "is_couponed",
            "buy_box_availability", "delivery_details", "sustainability_badges", "image_url", "product_url", "collected_at"]

df = pd.read_csv(ARCHIVO)
controles = {
    "huella MD5": (hashlib.md5(ARCHIVO.read_bytes()).hexdigest(), "402f2540bbbd4f7823b8e1309f970c46"),
    "tamaño (bytes)": (ARCHIVO.stat().st_size, 38_552_533),
    "filas": (len(df), 42_675),
    "columnas": (list(df.columns), COLUMNAS),
}
for nombre, (obtenido, esperado) in controles.items():
    estado = "OK     " if obtenido == esperado else "REVISAR"
    print(f"{estado} {nombre}: {obtenido if nombre != 'columnas' else f'{len(obtenido)} columnas'}")
```
